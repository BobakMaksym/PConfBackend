```python
from typing import Any


from bson import ObjectId
from pydantic import BaseModel
from uuid import uuid4
from pydantic.types import UUID4
from pydantic_extra_types.mongo_object_id import MongoObjectId
```

```python
class BaseUser(BaseModel):
    id: Any
    name: str

class UserModelInner(BaseUser):
    id: UUID4

class UserInn(BaseUser):
    id: MongoObjectId

class UserOut(BaseUser):
    id: str
```

# ObjectId

```python
user_inner = UserMongo(id=ObjectId(), name="max")
```

```python
user_inner
```

```python
user_inner_dict = user_inner.model_dump(mode="json")

user_out = UserModelOut.model_validate(user_inner_dict)

user_out
```

```python
user_inner = UserMongo.model_validate(user_out.model_dump())

user_inner
```

# UUID

```python
user_inner = UserModelInner(id=uuid4(), name="max")
```

```python
user_inner
```

```python
user_inner_dict = user_inner.model_dump(mode="json")

print(user_inner_dict)
print(type(user_inner_dict))
print(type(user_inner_dict["id"]))

user_out = UserModelOut.model_validate(user_inner_dict)
```

```python
user_out
```

```python
user_inner = UserModelInner.model_validate(user_out.model_dump())
user_inner
```

# UUID and ObjectId serialization

```python
printf = lambda obj: print(f"{type(obj).__name__}( {obj} )")

user = {"id": ObjectId(), "name": "max"}

user = UserInn(**user)
printf(user)

user = user.model_dump(mode="json")
printf(user)

user = UserOut.model_validate(user)
printf(user)

user = UserInn.model_validate(user.model_dump())
printf(user)
```
