```python
class BaseDocument(Document):
    meta = {
        "db_alias": "default",
        "abstract": True,
    }

class CpuModel(Document):
    manufacturer: str = fields.StringField()
    model: str = fields.StringField()
    socket: str = fields.StringField()
    ram_types: str = fields.StringField()
    igp: bool = fields.BooleanField(default=False)
    tdp: int = fields.IntField()
    tdp_max: int = fields.IntField()
    cores: int = fields.IntField()
    threads: int = fields.IntField()
    
    meta = {"collection": "cpu"}
```

```python
cpu = CpuModel(
    manufacturer="100", model="100", socket="100", ram_types="ddr4", igp=True, tdp=100, tdp_max=100, cores=100, threads=100,
)
cpu.save()
```

```python
cpus = CpuModel.objects()
for cpu in cpus:
    print(cpu)

print(len(cpus))
```
